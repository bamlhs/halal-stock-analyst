# Halal Stock Analyst

Runs a Sharia-compliance screener + a locally-loaded open-source LLM chat analyst, exposed publicly via an ngrok tunnel.

**Before running:** open the key icon (Secrets) in the left sidebar and add a secret named `NGROK_AUTHTOKEN` with your ngrok auth token. Never paste the token directly into a cell — Colab secrets keep it out of the notebook file and out of this public repo.

**Disclaimer:** this tool approximates common AAOIFI / Islamic-index screening methodology. It is not a fatwa. Confirm with a qualified scholar or recognized Islamic index before investing.

Runtime: Runtime → Change runtime type → GPU (T4 is enough for the default 7B model in 4-bit).

```
from google.colab import userdata
import os

# Pulled from Colab's encrypted secrets store, never hardcoded.
os.environ["NGROK_AUTHTOKEN"] = userdata.get("NGROK_AUTHTOKEN")
```

```
import threading
import uvicorn
from pyngrok import ngrok

from src.app import app

def run():
    uvicorn.run(app, host="0.0.0.0", port=8000, log_level="info")

thread = threading.Thread(target=run, daemon=True)
thread.start()

public_url = ngrok.connect(8000)
print(f"Public URL: {public_url}")
```

Open the printed `public_url` in a browser to use the chat UI, or share it with whoever you're giving access to.

The first `/chat` request will take a minute or two while the 7B model loads into GPU memory — subsequent requests are fast.
